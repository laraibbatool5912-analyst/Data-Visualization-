## 11. Correlation Analysis

Correlation is used to examine the strength and direction of linear relationships among numerical variables.

A correlation close to **+1** indicates a strong positive linear relationship, a value near **0** indicates a weak linear relationship, and a value close to **−1** indicates a strong negative linear relationship.

```python
numeric_columns = df.select_dtypes(include=np.number).columns
correlation = df[numeric_columns].corr()

display(correlation.round(3))
```

```python
plt.figure(figsize=(12, 8))
sns.heatmap(
    correlation,
    annot=True,
    cmap="coolwarm",
    fmt=".2f",
    linewidths=0.5
)
plt.title("Correlation Heatmap — World Happiness")
plt.tight_layout()
plt.show()
```

### Correlation with Happiness Score

The following table ranks the numerical variables by their correlation with Happiness Score. This provides a clearer answer to the question of which measured factors have the strongest linear association with happiness.

```python
happiness_corr = (
    correlation["Happiness_Score"]
    .drop("Happiness_Score")
    .sort_values(ascending=False)
    .to_frame("Correlation with Happiness Score")
)

display(happiness_corr.round(3))
```

## 12. GDP vs Happiness Score

The original analysis identified GDP as one of the strongest positive relationships with Happiness Score. A scatter plot and regression line are used to visualize this association.

```python
plt.figure(figsize=(10, 6))
sns.scatterplot(data=df, x="GDP", y="Happiness_Score")
sns.regplot(
    data=df,
    x="GDP",
    y="Happiness_Score",
    scatter=False
)
plt.title("GDP vs Happiness Score")
plt.xlabel("GDP per Capita")
plt.ylabel("Happiness Score")
plt.tight_layout()
plt.show()

print(f"Correlation between GDP and Happiness Score: "
      f"{df['GDP'].corr(df['Happiness_Score']):.3f}")
```

### Interpretation

If the regression line slopes upward, countries with higher GDP values tend to have higher Happiness Scores in this dataset. This is an **association**, not evidence that GDP alone causes happiness.

## 16. Regional Happiness Analysis

Average Happiness Score is calculated for each region to compare regional patterns.

```python
regional = (
    df.groupby("Region")["Happiness_Score"]
    .mean()
    .sort_values(ascending=False)
)

display(regional.to_frame("Average Happiness Score").round(3))

print("Highest-average region:", regional.index[0])
print(f"Highest regional average: {regional.iloc[0]:.3f}")
```

```python
plt.figure(figsize=(12, 6))
sns.barplot(
    x=regional.values,
    y=regional.index,
    orient="h"
)
plt.title("Average Happiness Score by Region")
plt.xlabel("Average Happiness Score")
plt.ylabel("Region")
plt.tight_layout()
plt.show()
```

## 17. Outlier Screening

A simple IQR-based method is used to flag potential outliers in selected numerical variables. This is a screening technique; flagged observations are not automatically errors.

```python
def iqr_outlier_summary(dataframe, columns):
    rows = []
    for col in columns:
        q1 = dataframe[col].quantile(0.25)
        q3 = dataframe[col].quantile(0.75)
        iqr = q3 - q1
        lower = q1 - 1.5 * iqr
        upper = q3 + 1.5 * iqr
        count = ((dataframe[col] < lower) | (dataframe[col] > upper)).sum()
        rows.append({
            "Variable": col,
            "Q1": q1,
            "Q3": q3,
            "IQR": iqr,
            "Lower Bound": lower,
            "Upper Bound": upper,
            "Potential Outliers": int(count)
        })
    return pd.DataFrame(rows)

outlier_columns = [
    col for col in [
        "Happiness_Score",
        "GDP",
        "Social_Support",
        "Life_Expectancy",
        "Freedom",
        "Generosity",
        "Corruption"
    ]
    if col in df.columns
]

outlier_summary = iqr_outlier_summary(df, outlier_columns)
display(outlier_summary.round(3))
```
